```python
# collection: list, tuple, set, dict
# List : ordered, mutable, allows duplicates
fruits = ['apple','mango','orange','pineapple']
fruits[2] = 'watermelon' # update the existing element
fruits.append('blueberry')
fruits.insert(2, 'banana')
fruits.extend(['kiwi','papaya','grapes'])
fruits.remove('apple')
fruits.sort(reverse=True)
print(fruits)
print(type(fruits))
print(len(fruits))
print(fruits[4])
print(fruits[-1]) # print last element from the list
print(fruits[1:3])
print(fruits[1:])
print(fruits[:3])
```

```python
# While loop
lucky_number = 5675
while True:
  number = int(input("Enter a lucky number: "))
  if number >=1000 and number <=9999:
    if number == lucky_number:
      print("congratulations! You have guessed the number")
      break
    else:
      print("try again!")
  else:
    print("Enter any 4-digit number")
```

```python
# take two inputs from the user and if the products is greater than 5000 print "Pass" else keep trying.
while True:
  a = int(input("Enter number a: "))
  b = int(input("Enter number b: "))
  if a >0 and b >0:
    if a*b > 5000:
      print("pass")
      break
    else:
      print("try again!")
  else:
    print("Enter any positive number")
```
